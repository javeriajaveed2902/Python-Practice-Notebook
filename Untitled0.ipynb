 Build a program to create a movie watchlist. Ask the user to input two movie names
(str), their ratings (float), and genres (str). Store each movie’s details in a dictionary (name,
rating, genre) and put both dictionaries in a list. Use a tuple to store possible streaming
platforms (e.g., "Netflix", "Hulu", "Disney+"). Add the genres to a set to track unique genres.
Display the watchlist, the platforms tuple, the unique genres set, and the average rating of
the two movies (float)

```python
movie_1=input("enter 1st movie name:")
rating_1=float(input("enter 1st movie rating:"))
genre_1=input("enter 1st genre:")
movie_2=input("enter 2nd movie name:")
rating_2=float(input("enter 2nd movie rating:"))
genre_2=input("enter 2nd genre:")
```

```python
dict_1={
    "name":movie_1,
    "rating":rating_1,
    "genre":genre_1
}
dict_2={
    "name":movie_2,
    "rating":rating_2,
    "genre":genre_2
}
list_of_dicts=[]
list_of_dicts.append(dict_1)
list_of_dicts.append(dict_2)
```

```python
plateforms=("Netflix", "Hulu", "Disney+")

unique_genres=set()
unique_genres.add("action")
unique_genres.add("thriller")
average=(rating_1+rating_2)/2
print(list_of_dicts)
print(plateforms)
print(unique_genres)
print(average)


```
